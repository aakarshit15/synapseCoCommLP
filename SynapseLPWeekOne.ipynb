###Armstrong number

Armstrong number is a number that is equal to the sum of the nth power of its digits. Here, n is the number of digits of the Armstrong number.

For example 0, 1, 153, 370, 371 and 407 are the Armstrong numbers.

Say, 153.
1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153!
Armstrong.

````python
# Follow good practices and write code that accepts start and limit integers,
# and returns a list of Armstrong numbers between them (both inclusive).

# For calculating sum of digits raised to number of digits
def armstrongSum(num):
  sum = 0
  temp = num
  while temp != 0:
    s = (temp % 10) ** len(str(num))
    sum += s
    temp //= 10
  return sum

# To check if armstrong or not
def isArmstrong(num):
  return num == armstrongSum(num)

# To iterate over a list for checking armstrong or not
def arms(start, end):
  armList = []
  for num in range(start, end+1):
    if isArmstrong(num):
      armList.append(num)
  return armList

print(arms(1, 500))
````

###Collatz sequence

Given a number, if it is even, the next number in collatz sequence is n/2. If it is odd, the next number is 3*n + 1.

If you do this enough times, you will DEFINITELY end with a sequence of 4, 2, 1.

(Search up how later, confusing but interesting)

Return the collatz sequence of the input number, ending with the 4, 2, 1.

````python
# Taking one input number, return its Collatz sequence in a list.
# Ending with the 4, 2, 1
def coaltzSeq(num):
  seq = []
  while num >= 1:
    seq.append(num)
    if num % 2 == 0:
      num /= 2
    else:
      if num == 1:
        break
      else:
        num = 3*num + 1
  return seq

coaltzSeq(10)
````

To actually understand what this means, we first need to know what all this crab jargon actually is. So, on Googling, you will discover the following meanings:

*   Weight - total teight
*   Shucked - weight of meat only
*   Viscera - gut weight, after bleeding

Well, those are some pretty disgusting and gory definitions.

Moving past that, we here find that some weight statistics is lost. What about the other potential body parts of a crab? Maybe we need them to accurately predict its (apparently very short) lifespan? Maybe it is actually needed in some future model we choose to implement?

So, make a new column called "Lost Weight". Use the following formula to populate every row of it:



```
Weight - (Shucked + Viscera + Shell)
```


If this operation gives a value of Lost Weight that is less than zero, then assign zero to that row's Lost Weight value. Otherwise, assign one to it.

Also, ensure that this column is added in between Shell Weight and Age.

````python
# write code here
df["Lost Weight"] = df["Weight"] - (df["Shucked Weight"] + df["Viscera Weight"] + df["Shell Weight"])
df.loc[df["Lost Weight"] < 0, "Lost Weight"] = 0
lost_weight_col = df.pop("Lost Weight")
df.insert(8, "Lost Weight", lost_weight_col)
df.head()
````

For standardisation, bear in mind that the one hot encoded columns are not standardised. They are categorical in nature, so it makes no sense to shoehorn them into any type of distribution at all.

But, its ok to normalise them, as it will be scaled down to values between 0 and 1, which is the range in which they already exist anyways. Normalisation doesn't affect them.

Standardise df_std such that the one hot label encoded columns aren't affected.

````python
from sklearn.preprocessing import StandardScaler
````

````python
# write code here
df_std = dummy_df.copy() # copying data frame
std_cols = ["Length", "Diameter", "Height", "Weight", "Shucked Weight", "Viscera Weight", "Shell Weight", "Lost Weight", "Age"] # list of columns that need to be standarized
for col in std_cols: # for all columns in std_cols
  std_scaler = StandardScaler().fit(df_std[[col]]) # fitting it to make a scale
  df_std[col] = std_scaler.transform(df_std[[col]]) # transforming data according to scale
df_std = pd.DataFrame(df_std) # converting the scaled data to data frame
for (std_col, col) in zip(df_std.columns, dummy_df.columns): # renaming columns
  df_std.rename(columns={std_col: col}, inplace=True)
df_std.head()
````

Boxplots are used to visualize the distribution of data in a datasets. The data is divided into quartiles. And there are whiskers i.e. lines present before and after the boxes. And hence sometimes it is called as box and whisker plot also. This whisker is generally the 1.5 times the length of box i.e. IQR (Inter Quartile Range) b/w Q1 and Q3 i.e. IQR = Q3 - Q1.

The data in box plot is mostly divided between 5 parts:
1. **Minimum**: The smallest point within the range of whiskers (generally, Q1 - 1.IQR)
2. **1st Quartile(Q1)**: The 1st quartile, which is the 25th percentile of the data. This means that 25% of the data points are below this value.
3. **2nd Quartile(Q2)**: It represents the median by the line inside the box. It divides the dataset into two equal halves
4. **3rd Quartile(Q3)**: The third quartile, which is the 75th percentile. This means that 75% of the data points are below this value.
5. **Maximum**: The highest point within the range of whiskers (generally, Q3 + 1.IQR)

Apart from this there are outliers. They are represented by dots, diamonds, etc. They lie before and after the whiskers.

The boxplot is mostly used to see the distribution of data. The line of median in the box, the length of whiskers and the number of outliers helps us to see if the data is symmetric or positively skewed or negatively skewed.
